```python

# Exploring CIA Triad with Simulations
# Demonstrates Confidentiality, Integrity, and Availability

import hashlib
import hmac
import random
import time
from cryptography.fernet import Fernet

# ================== CONFIDENTIALITY ==================
print("=== 🔒 CONFIDENTIALITY DEMO ===")

# Generate encryption key & cipher
secret_key = Fernet.generate_key()
cipher_tool = Fernet(secret_key)

# Sensitive data
message = b"Confidential Contract: Payment = $11,000,000"

# Encrypt & Decrypt
cipher_text = cipher_tool.encrypt(message)
plain_text = cipher_tool.decrypt(cipher_text).decode()

print("Encryption Key :", secret_key.decode()[:25] + "...")   # show partial key
print("Cipher Text    :", cipher_text[:60], "...")            # shortened display
print("Decrypted Text :", plain_text)


# ================== INTEGRITY ==================
print("\n=== ✅ INTEGRITY DEMO ===")

# Compute SHA-256 hash of original & tampered data
hash_original = hashlib.sha256(message).hexdigest()
tampered_msg = b"Confidential Contract: Payment = $9,000,000"
hash_tampered = hashlib.sha256(tampered_msg).hexdigest()

print("Original Hash  :", hash_original)
print("Tampered Hash  :", hash_tampered)

# Check integrity using hmac.compare_digest
if hmac.compare_digest(hash_original, hash_tampered):
    print("Result: ✔ Data is authentic")
else:
    print("Result: ⚠ Data has been tampered!")


# ================== AVAILABILITY ==================
print("\n=== 🚦 AVAILABILITY DEMO ===")

def server_request():
    """Simulate server with ~75% uptime probability"""
    if random.random() < 0.75:
        return "Server Response: ✅ Data Available"
    else:
        raise ConnectionError("Server ❌ Down")

# Retry mechanism (simulate multiple attempts)
for attempt in range(1, 6):
    try:
        reply = server_request()
        print(f"Attempt {attempt}: {reply}")
        break
    except ConnectionError as err:
        print(f"Attempt {attempt}: {err}, retrying...")
        time.sleep(1)

```

```python
# 🔐 CIA Triad Simulation: Confidentiality | Integrity | Availability
import hashlib, hmac, random, time
from cryptography.fernet import Fernet

# ---------- CONFIDENTIALITY ----------
key = Fernet.generate_key()
cipher = Fernet(key)
msg = b"Confidential: Payment=$11,000,000"
ct = cipher.encrypt(msg)
print("🔒 Confidentiality\nCipher:", ct[:50], "...\nDecrypted:", cipher.decrypt(ct).decode())

# ---------- INTEGRITY ----------
print("\n✅ Integrity")
h1 = hashlib.sha256(msg).hexdigest()
h2 = hashlib.sha256(b"Confidential: Payment=$9,000,000").hexdigest()
print("Original Hash:", h1[:20], "...\nTampered Hash:", h2[:20], "...")
print("Status:", "Authentic ✔" if hmac.compare_digest(h1, h2) else "Tampered ⚠")

# ---------- AVAILABILITY ----------
print("\n🚦 Availability")
for i in range(1, 6):
    try:
        if random.random() < 0.75:
            print(f"Attempt {i}: ✅ Server Available")
            break
        raise ConnectionError("❌ Down")
    except:
        print(f"Attempt {i}: Retry..."); time.sleep(0.5)
```
